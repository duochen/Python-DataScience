```python
import matplotlib.pyplot as plt
```

```python
x = [0, 1, 2, 3, 4, 5]
y1 = [0, 1, 4, 9, 16, 25]   # Quadratic
y2 = [0, 1, 2, 3, 4, 5]     # Linear

# Create plot
plt.plot(x, y1, label="Quadratic", lw=3)   # thicker line
plt.plot(x, y2, label="Linear", linewidth=1)      # thinner line

# Add labels, title, and legend
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Matplotlib Plot with Different Line Widths")
plt.legend()

# Show the plot
plt.show()
```
